# FEATURE ENGINEERING & EXTRACTION

```python
df_cleaned['bmi_category']=pd.cut(
    df_cleaned['bmi'],
    bins=[0,18.5,24.9,29.9,float('inf')],
    labels=['underweight','normlal','overweight','obese']
    
)
```

### here we get the bmi category it is in str so we have to convert it into text for ml work through one hot encoding

```python
df_cleaned=pd.get_dummies(df_cleaned,columns=['bmi_category'],drop_first=True)
```

```python
df_cleaned=df_cleaned.astype(int)
```

```python
df_cleaned.head()
```

```python
from sklearn.preprocessing import StandardScaler
cols=['age','bmi','children']
scaler = StandardScaler()
df_cleaned[cols]=scaler.fit_transform(df_cleaned[cols])
```

```python
df_cleaned
```

```python
from scipy.stats import pearsonr

# ----------------------------------
# Pearson Correlation Calculation
# ----------------------------------

# List of features to check against target
selected_features = [
    'age', 'bmi', 'children', 'is_female', 'is_smoker',
    'region_northwest', 'region_southeast', 'region_southwest',
    'bmi_category_normlal', 'bmi_category_overweight', 'bmi_category_obese'
]

correlations = {
    feature: pearsonr(df_cleaned[feature], df_cleaned['charges'])[0]
    for feature in selected_features
}
correlation_df = pd.DataFrame(list(correlations.items()), columns=['Feature', 'Pearson Correlation'])
correlation_df.sort_values(by='Pearson Correlation', ascending=False)
```

```python
cat_features = [
    'is_female', 'is_smoker',
    'region_northwest', 'region_southeast', 'region_southwest',
    'bmi_category_normlal', 'bmi_category_overweight', 'bmi_category_obese'
]
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05

df_cleaned['charges_bin'] = pd.qcut(df_cleaned['charges'], q=4, labels=False)
chi2_results = {}

for col in cat_features:
    contingency = pd.crosstab(df_cleaned[col], df_cleaned['charges_bin'])
    chi2_stat, p_val, _, _ = chi2_contingency(contingency)
    decision = 'Reject Null (Keep Feature)' if p_val < alpha else 'Accept Null (Drop Feature)'
    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }

chi2_df = pd.DataFrame(chi2_results).T
chi2_df = chi2_df.sort_values(by='p_value')
chi2_df
```

# MODEL CREATION

```python
from sklearn.model_selection import train_test_split
```

```python
x=final_df.drop('charges',axis=1)
y=final_df['charges']
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression 
```

```python
model=LinearRegression ()
model.fit(X_train,y_train)
```

# NOW PREDICTION WORKS HERE 
- WE PREDICT THE  X TEST FROM THIS WE GET THE Y PREDICT
- NOW WE COMPARE THE Y PREDICT WITH Y TEST 

```python
y_pred=model.predict(X_test)
```

```python
y_pred
```

# NOW WE WILLSEE HOW WELL OUR MODEL PERFORMED

```python
from sklearn.metrics import r2_score
```

```python
r2=r2_score(y_test,y_pred)
r2
```

## model trained with 80 percent

# ADJUSTED R SQR

```python
n=X_test.shape[0]
p=X_test.shape[1]
adjusted_r2=1-((1-r2)*(n-1)/(n-1-p))


adjusted_r2
```
